```python
import pandas as pd
df = pd.read_csv('claims_main.csv')
```

```python
df.shape
```

```python
payers_df = df[['payer_type']].drop_duplicates()
payers_df
```

```python
payers_df['payer_id'] = range(1, len(payers_df) + 1)
```

```python
payers_df
```

```python
denial_reasons_df = df[['denial_reason_code', 'denial_category']].drop_duplicates()
denial_reasons_df
```

```python
denial_reasons_df['denial_id'] = range(1, len(denial_reasons_df) + 1)
denial_reasons_df
```

```python
claims_df = df.merge(payers_df, on='payer_type')
claims_df = claims_df.merge(denial_reasons_df, on=['denial_reason_code', 'denial_category'])
claims_df.shape
```

```python
import sqlite3
conn = sqlite3.connect('claims.db')
```

```python
claims_df.to_sql('claims', conn, if_exists='replace', index=False)
```

```python
payers_df.to_sql('payers', conn, if_exists='replace', index=False)
```

```python
denial_reasons_df.to_sql('denial_reasons', conn, if_exists='replace', index=False)
```
